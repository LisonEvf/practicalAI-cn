```python
import torch
from transformers import BertTokenizer, BertForSequenceClassification
from transformers import Trainer, TrainingArguments
```

# BERT 简介 Introduction to BERT

```python
# 加载预训练模型 Load pretrained model
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
model = BertForSequenceClassification.from_pretrained('bert-base-uncased', num_labels=2)

print("BERT Model loaded successfully!")
print(f"Model parameters: {sum(p.numel() for p in model.parameters()):,}")
```

```python
# 文本编码 Text encoding
text = "This is an amazing example of natural language processing!"
inputs = tokenizer(text, return_tensors='pt', padding=True, truncation=True)

print("Input IDs:")
print(inputs['input_ids'])
print("\nAttention Mask:")
print(inputs['attention_mask'])
```

```python
# 前向传播 Forward pass
with torch.no_grad():
    outputs = model(**inputs)
    logits = outputs.logits
    print(f"Logits shape: {logits.shape}")
    print(f"Logits: {logits}")
```

# 情感分类示例 Sentiment Classification Example

```python
# 使用 BERT 进行情感分类
sentences = [
    "I absolutely love this product!",
    "This is the worst experience ever.",
    "It's okay, nothing special."
]

for sent in sentences:
    inputs = tokenizer(sent, return_tensors='pt', padding=True, truncation=True)
    with torch.no_grad():
        outputs = model(**inputs)
        probs = torch.softmax(outputs.logits, dim=1)
        pred = torch.argmax(probs, dim=1)
    label = "Positive" if pred.item() == 1 else "Negative"
    print(f"Sentence: '{sent}'")
    print(f"Prediction: {label} (prob: {probs[0][pred].item():.4f})\n")
```
