主题建模是一种无监督学习方法，用于发现文档集合中潜在的主题。本notebook介绍如何使用 LDA（潜在狄利克雷分配）进行主题建模。

Topic modeling is an unsupervised learning method for discovering latent topics in document collections. This notebook introduces how to use LDA (Latent Dirichlet Allocation) for topic modeling.

<img src="https://raw.githubusercontent.com/LisonEvf/practicalAI-cn/master/images/topic.png" width=500>

# 概述 Overview

* **目标:**  发现文档集合中的潜在主题结构。
* **优点:** 
  * 无需标注数据
  * 可揭示文本语义结构
  * 用于文档聚类和摘要
* **缺点:**
  * 需要选择主题数量
  * 训练可能慢
* **其他:** 
  * LDA 是最经典的主题模型
  * 有 Neural LDA、CTM 等变体

# 设置 Setup

```python
import numpy as np
import pandas as pd
import re
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

# 数据准备 Data Preparation

```python
# 示例文档 Example documents
documents = [
    "Sports are great for physical and mental health",
    "The economy is showing signs of improvement this quarter",
    "Scientists discover new species in the Amazon rainforest",
    "The latest smartphone technology features amazing camera",
    "Football players train hard for the championship game",
    "Stock market trends indicate potential growth in tech sector",
    "New medical research shows benefits of exercise",
    "Music festivals attract thousands of fans every year",
    "Political debates focus on healthcare and education",
    "Climate change affects global weather patterns"
]
```

```python
# 文本预处理 Text preprocessing
def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    return text

preprocessed_docs = [preprocess_text(doc) for doc in documents]
print("Preprocessed documents:")
for i, doc in enumerate(preprocessed_docs):
    print(f"{i}: {doc}")
```

# 创建文档-词矩阵 Create Document-Term Matrix

```python
# 使用 CountVectorizer 创建词袋模型
vectorizer = CountVectorizer(max_df=0.9, min_df=1, stop_words='english')
doc_term_matrix = vectorizer.fit_transform(preprocessed_docs)

print(f"Document-term matrix shape: {doc_term_matrix.shape}")
print(f"Vocabulary size: {len(vectorizer.vocabulary_)}")
```

# LDA 模型 LDA Model

```python
# 训练 LDA 模型 Train LDA model
num_topics = 3
lda_model = LatentDirichletAllocation(
    n_components=num_topics,
    max_iter=10,
    learning_method='online',
    random_state=42
)
lda_model.fit(doc_term_matrix)

# 获取词表 Get feature names
feature_names = vectorizer.get_feature_names_out()
```

```python
# 显示每个主题的关键词 Display topic keywords
def display_topics(model, feature_names, n_top_words=5):
    for topic_idx, topic in enumerate(model.components_):
        top_words_idx = topic.argsort()[:-n_top_words - 1:-1]
        top_words = [feature_names[i] for i in top_words_idx]
        print(f"Topic {topic_idx}: {', '.join(top_words)}")

print("Topics discovered:")
display_topics(lda_model, feature_names)
```

```python
# 文档的主题分布 Document topic distribution
doc_topics = lda_model.transform(doc_term_matrix)
print("\nDocument topic distribution:")
for i, doc in enumerate(documents):
    topic_idx = np.argmax(doc_topics[i])
    print(f"Doc {i}: Topic {topic_idx} (prob: {doc_topics[i][topic_idx]:.3f})")
```
